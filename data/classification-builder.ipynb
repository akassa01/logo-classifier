```python
def normalize_data(df: pd.DataFrame, industry_col: str = "industry") -> pd.DataFrame:
    df = df.copy()
    df[industry_col] = df[industry_col].replace(INDUSTRY_RENAMES)
    df["sector"] = df[industry_col].map(INDUSTRY_TO_SECTOR).fillna("Other")
    return df
```

```python
# dataset labeling code
raw = pd.read_csv(raw_ds, usecols=["name", "domain", "industry", "size range"], low_memory=False)
raw = raw.dropna(subset=["domain", "industry"])
raw = raw[raw["size range"] != "1 - 10"]
raw = raw[~raw["industry"].str.lower().isin({"law practice", "legal services", "alternative dispute resolution"})] # 21 classification adjustment

raw = normalize_data(raw, "industry")

labeled = logos.merge(raw[["name", "domain", "sector"]], on="domain", how="inner")

labeled.to_csv("/kaggle/working/labels.csv")
```
